## Fase 1 · El Artesano (15 min, sin IA)

Escribe a mano la implementación en `artesano_sesion10_c.c`. El archivo ya trae la

````python
%%writefile api_sesion10.h
/* api_sesion10.h
 * Sesion 10 - Mini-evaluador de expresiones
 * Contrato compartido por el Artesano y el Cientifico.
 */

#ifndef API_SESION10_H
#define API_SESION10_H

/* error = 0: expresion valida
 * error = 1: division entre cero
 * error = 2: sintaxis invalida
 */
long evaluar_expresion(const char *fuente, int *error);

#endif /* API_SESION10_H */
````

````python

%%writefile ia_sesion10_c.c
#include "api_sesion10.h"
#include <ctype.h>
#include <stddef.h>

typedef struct {
    const char *entrada;
    int error;
} Analizador;

static void omitir_espacios(Analizador *a) {
    while (isspace((unsigned char)*a->entrada))
        a->entrada++;
}

static long analizar_expresion(Analizador *a);

static long analizar_numero(Analizador *a) {
    omitir_espacios(a);

    if (*a->entrada == '(') {
        a->entrada++;

        long resultado = analizar_expresion(a);

        if (a->error)
            return 0;

        omitir_espacios(a);

        if (*a->entrada != ')') {
            a->error = 2;
            return 0;
        }

        a->entrada++;
        return resultado;
    }

    if (!isdigit((unsigned char)*a->entrada)) {
        a->error = 2;
        return 0;
    }

    long numero = 0;

    while (isdigit((unsigned char)*a->entrada)) {
        numero = numero * 10 + (*a->entrada - '0');
        a->entrada++;
    }

    return numero;
}

static long analizar_producto(Analizador *a) {
    long resultado = analizar_numero(a);

    while (!a->error) {
        omitir_espacios(a);

        char operador = *a->entrada;

        if (operador != '*' && operador != '/')
            break;

        a->entrada++;

        long valor = analizar_numero(a);

        if (a->error)
            return 0;

        if (operador == '*') {
            resultado *= valor;
        } else {
            if (valor == 0) {
                a->error = 1;
                return 0;
            }

            resultado /= valor;
        }
    }

    return resultado;
}

static long analizar_expresion(Analizador *a) {
    long resultado = analizar_producto(a);

    while (!a->error) {
        omitir_espacios(a);

        char operador = *a->entrada;

        if (operador != '+' && operador != '-')
            break;

        a->entrada++;

        long valor = analizar_producto(a);

        if (a->error)
            return 0;

        if (operador == '+')
            resultado += valor;
        else
            resultado -= valor;
    }

    return resultado;
}

long evaluar_expresion(const char *fuente, int *error) {
    if (error != NULL)
        *error = 0;

    if (fuente == NULL) {
        if (error != NULL)
            *error = 2;
        return 0;
    }

    Analizador a = {fuente, 0};

    long resultado = analizar_expresion(&a);

    omitir_espacios(&a);

    if (!a.error && *a.entrada != '\0')
        a.error = 2;

    if (error != NULL)
        *error = a.error;

    return a.error ? 0 : resultado;
}
````

## Fase 2 · El Científico (10 min)

Redacta el contrato **C-R-E-O** (Contexto, Rol, Especificación, Output) para pedirle
la misma función a la IA, y pega el resultado en la celda de `ia_sesion10_c.c`. La
firma debe ser idéntica a la del contrato; si la IA la cambia, corrígela antes de
compilar.


````python

%%writefile artesano_sesion10_c.c
#include "api_sesion10.h"
#include <ctype.h>
#include <stddef.h>

typedef struct {
    const char *s;
    int error;
} Parser;

static void saltar(Parser *p) {
    while (*p->s == ' ')
        p->s++;
}

static long expr(Parser *p);

static long primaria(Parser *p) {
    saltar(p);

    if (*p->s == '(') {
        p->s++;

        long valor = expr(p);

        if (p->error)
            return 0;

        saltar(p);

        if (*p->s != ')') {
            p->error = 2;
            return 0;
        }

        p->s++;
        return valor;
    }

    if (!isdigit((unsigned char)*p->s)) {
        p->error = 2;
        return 0;
    }

    long numero = 0;

    while (isdigit((unsigned char)*p->s)) {
        numero = numero * 10 + (*p->s - '0');
        p->s++;
    }

    return numero;
}

static long termino(Parser *p) {
    long resultado = primaria(p);

    while (!p->error) {
        saltar(p);

        char op = *p->s;

        if (op != '*' && op != '/')
            break;

        p->s++;

        long valor = primaria(p);

        if (p->error)
            return 0;

        if (op == '*') {
            resultado *= valor;
        } else {
            if (valor == 0) {
                p->error = 1;
                return 0;
            }

            resultado /= valor;
        }
    }

    return resultado;
}

static long expr(Parser *p) {
    long resultado = termino(p);

    while (!p->error) {
        saltar(p);

        char op = *p->s;

        if (op != '+' && op != '-')
            break;

        p->s++;

        long valor = termino(p);

        if (p->error)
            return 0;

        if (op == '+')
            resultado += valor;
        else
            resultado -= valor;
    }

    return resultado;
}

long evaluar_expresion(const char *fuente, int *error) {
    static int auxiliar;

    if (error == NULL)
        error = &auxiliar;

    *error = 0;

    if (fuente == NULL) {
        *error = 2;
        return 0;
    }

    Parser p = {fuente, 0};

    long resultado = expr(&p);

    saltar(&p);

    if (!p.error && *p.s != '\0')
        p.error = 2;

    *error = p.error;

    return p.error ? 0 : resultado;
}
````

## Fase 3 · El Auditor (10 min)

El harness de abajo **no aprueba nada por sí solo**: ejecuta tu función caso por caso,
la compara contra el valor esperado e imprime `OK` o `FALLA`. Léelo antes de ejecutarlo:
los casos te dicen qué se está midiendo.


````python
%%writefile harness_sesion10_c.c
/* harness_sesion10_c.c - auditoria automatica de la Sesion 10
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion10_c.c artesano_sesion10_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion10.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_int(const char *caso, long esperado, long obtenido) {
    int ok = (esperado == obtenido);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12ld obtenido: %-12ld %s\n",
           probadas, caso, esperado, obtenido, ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 10 - Mini-evaluador de expresiones (Hito 1) ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { int e=0; long r=evaluar_expresion("2+3*4",&e); (void)e; verificar_int("\"2+3*4\"", (long)(14), (long)(r)); }
    { int e=0; long r=evaluar_expresion("(2+3)*4",&e); (void)e; verificar_int("\"(2+3)*4\"", (long)(20), (long)(r)); }
    { int e=0; long r=evaluar_expresion("10/0",&e); (void)r; verificar_int("\"10/0\" -> error 1", (long)(1), (long)(e)); }
    { int e=0; long r=evaluar_expresion("2+",&e); (void)r; verificar_int("\"2+\" -> error 2", (long)(2), (long)(e)); }
    { int e=0; long r=evaluar_expresion("  7 ",&e); (void)e; verificar_int("\"  7 \"", (long)(7), (long)(r)); }
    { int e=0; long r=evaluar_expresion("5-10",&e); (void)e; verificar_int("\"5-10\"", (long)(-5), (long)(r)); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion10.md" "\n");
    return 0;
}
````
